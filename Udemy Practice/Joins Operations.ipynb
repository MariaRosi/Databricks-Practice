```python
from pyspark.sql.functions import col

df_inner = (emp_df.alias("e")
           .join(dept_df.alias("d"), col("e.emp_dept_id") == col("d.dept_id"), "inner")
           .select(col("e.emp_id").alias("Employee Id"), col("e.emp_name").alias("Employee Name"), col("d.dept_name").alias("Department Name"))
           )

display(df_inner)
```

```python
df_left = (emp_df.alias("e")
           .join(dept_df.alias("d"), col("e.emp_dept_id") == col("d.dept_id"), "left")
           .select(col("e.emp_id").alias("Employee Id"), col("e.emp_name").alias("Employee Name"), col("d.dept_name").alias("Department Name"))
           )

display(df_left)
```
