```python
data_1 = [
    [1, "Alice",   "Engineering", 75000],
    [2, "Bob",     "Finance",     82000],
    [3, "Charlie", "HR",          65000],
    [4, "Diana",   "Engineering", 78000]
]

data_2 = [
    [3, "Charlie", "HR",          65000],  # duplicate row
    [5, "Ethan",   "Marketing",   88000],
    [6, "Fiona",   "Finance",     79000],
    [7, "George",  "Engineering", 90000]
]

columns = ["id", "name", "department", "salary"]

df1 = spark.createDataFrame(data_1, columns)
df2 = spark.createDataFrame(data_2, columns)

display(df1)
display(df2)

```

```python
display(df1.union(df2))
```

```python
display(df1.union(df2).distinct())
```
