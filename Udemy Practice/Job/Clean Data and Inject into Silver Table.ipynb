```python
df = spark.sql("select * from udemy_practice_catalog.bronze.raw_sales_orders")

df.display()
```

```python
from pyspark.sql.functions import col, sum

df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df.columns
]).display()
```

```python
from pyspark.sql.functions import date_format, to_date

df_order_date_formatted = df.withColumn("order_date", to_date(date_format("order_date", "dd-MM-yyyy"), "dd-MM-yyyy"))

df_order_date_formatted.display()
```

```python
df_amount_decimal_converted = df_order_date_formatted.withColumn("amount", col("amount").cast("decimal(12,2)"))
df_amount_decimal_converted.display()
```

```python
df_amount_decimal_converted.printSchema()
```

```python
df_amount_decimal_converted.write.mode("overwrite").saveAsTable("udemy_practice_catalog.silver.clean_sales_orders")
```
