### **distinct And dropDuplicates**

```python

data = [
    (1, "Alice", 25),
    (2, "Bob", 30),
    (3, "Charlie", 35),
    (4, " David", 40),
    (5, "Eve", 28),
    (2, "Bob", 30),           # duplicate row
    (3, "Charlie", 35),       # duplicate row
    (6, "Alice", 25),         # duplicate values in 'name' and 'age'
    (7, "Eve", 28)            # duplicate values in 'name' and 'age'
]

header = ["id", "name", "age"]

df = spark.createDataFrame(data, header)

display(df)
```

```python
display(df.distinct())
```

```python
display(df.dropDuplicates())
```

```python
display(df.dropDuplicates(['name', 'age']))
```

```python
display(df.dropDuplicates(['name']))
```
