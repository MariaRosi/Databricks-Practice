```python
data = [
    [1,  "Alice",   28, "CA", 75000],
    [2,  "Bob",     34, "NY", 82000],
    [3,  "Charlie", 41, "TX", 95000],
    [4,  "Diana",   25, "CA", 68000],
    [5,  "Ethan",   37, "FL", 88000],
    [6,  "Fiona",   34, "NY", 79000],
    [7,  "George",  41, "TX", 102000],
    [8,  "Hannah",  28, "CA", 72000],
    [9,  "Ian",     25, "FL", 65000],
    [10, "Jane",    37, "NY", 90000]
]

columns = ["id", "name", "age", "state", "salary"]

df = spark.createDataFrame(data, columns)

display(df)
```

```python
display(df.groupBy("state").count())
```

```python
from pyspark.sql.functions import avg, sum, max

display(df.groupBy("state").agg(avg("salary").alias("avg_salary"), sum("salary").alias("total_salary"), max("salary").alias("max_salary")))
```
