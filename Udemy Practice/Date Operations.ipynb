```python
data = [
    (1, "Alice",   "2024-01-05", "2024-01-05 10:15:30"),
    (2, "Bob",     "2024-02-14", "2024-02-14 18:45:00"),
    (3, "Charlie", "2024-03-25", "2024-03-25 09:00:10"),
    (4, "Diana",   "2024-07-04", "2024-07-04 23:59:59"),
    (5, "Edward",  "2025-01-01", "2025-01-01 00:00:01")
]

columns = ["id", "name", "date", "timestamp"]

df = spark.createDataFrame(data, columns)

df.display()
```

```python

from pyspark.sql.functions import date_format

df.select("date", date_format("date", 'dd/MM/yyyy')).display()
```

```python
df.select("date", date_format("date", 'dd/MMM/yyyy')).display()
```

```python
df.select("date", date_format("date", 'dd/MMMM/yyyy')).display()
```

```python
df.select("timestamp", date_format("timestamp", 'dd-MMM-yyyy HH:mm:ss')).display()
```

```python
from pyspark.sql.functions import *

df.select("date", date_add("date", 5).alias("date_add")).display()
```

```python
df.select("date", date_sub("date", 5).alias("date_sub")).display()
```

```python
df1 = df.select("date", add_months("date", 2).alias("months_added"))

df1.display()
```

```python
df1.select("date", "months_added", datediff("months_added", "date").alias("day_difference")).display()
```

```python
df.select("date", current_date(), next_day("date", "Monday"), last_day("date")).display()
```

```python
df.select("date", year("date"), quarter("date"), month("date"), day("date")).display()
```
