```python
data = [
    [1, "Alice",   28, "CA", 75000],
    [2, "Bob",     None, "NY", 82000],
    [3, None,      41, "TX", None],
    [4, "Diana",   25, None, 68000],
    [5, "Ethan",   None, "FL", 88000],
    [6, None,      34, None, None],
    [7, "George",  41, "TX", 102000],
    [8, "Hannah",  28, "CA", None],
    [9, "Ian",     None, None, 65000],
    [10, None,     None, "NY", None]
]

cols = ["id", "name", "age", "state", "salary"]

df = spark.createDataFrame(data, cols)

display(df)
```

### **na.fill / fillna -  it replaces the null value based upon the filling value provided**
### **If the filling value is interger then all the integer column null value will be replaced**
### **If the filling value is string then only the string column null values will be replaced**



```python
display(df.na.fill("unknown"))
```

```python
display(df.fillna(0))
```

```python
display(df.fillna("Unknown", ["state", "name"]).fillna(0, ["salary", "age"]))
```
