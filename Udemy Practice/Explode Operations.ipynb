```python
data = [
    (1, "Alice",   ["Laptop", "Mouse", "Keyboard"], {"UK": 1200, "EU": 1300}),
    (2, "Bob",     ["Phone", "Charger"],            {"UK": 800}),
    (3, "Charlie", ["Tablet"],                      {"US": 600, "UK": 650}),
    (4, "Diana",   [],                              {"EU": 900}),
    (5, "Edward",  None,                            None)
]

columns = ["id", "name", "products", "sales"]
df = spark.createDataFrame(data, columns)
df.display()
```

```python
from pyspark.sql.functions import explode, posexplode

df.select("id", "name", explode("products")).display()
```

```python
df.select("id", "name", explode("products"), explode("sales")).display()
```

```python
df.select("id", "name", posexplode("products"), posexplode("sales")).display()
```
