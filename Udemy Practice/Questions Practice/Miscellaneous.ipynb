```python
%python


data = [
    ("C001","John","TXN1001","2024-01-10",250,"London"),
    ("C002","Alice","TXN1002","2024-01-11",300,"Manchester"),
    ("C001","John","TXN1003","2024-01-15",150,"London"),
    ("C003","Bob","TXN1004","2024-01-20",400,"Birmingham"),
    ("C001","John","TXN1003","2024-01-15",150,"London"),
    ("C004","Carol","TXN1005","2024-01-22",500,"Leeds"),
    ("C002","Alice","TXN1006","2024-02-01",200,"Manchester"),
    ("C005","David","TXN1007","2024-02-05",350,"London"),
    ("C002","Alice","TXN1006","2024-02-01",200,"Manchester"),
    ("C006","Eva","TXN1008","2024-02-10",450,"Bristol")
]

columns = [
    "customer_id",
    "customer_name",
    "transaction_id",
    "transaction_date",
    "amount",
    "city"
]


df = spark.createDataFrame(data, columns)
df.show()
```

```python
%python
df.distinct().show()
```

```python
(
    df
    .groupBy(df.columns)
    .count().alias("count")
    .filter("count > 1")
).show()
```

```python
from pyspark.sql.functions import col

df.dropDuplicates(["customer_id"]).show()
```

```python
df.dropDuplicates(["customer_id", "transaction_id"]).orderBy("customer_id").show()
```

```python
# first drop the duplicates
from pyspark.sql.functions import count

df_final = (
            df
            .distinct()
            .groupBy("customer_id")
            .agg(count("customer_id").alias("count"))
            .filter("count > 1")
        )

df_final.show()
```
