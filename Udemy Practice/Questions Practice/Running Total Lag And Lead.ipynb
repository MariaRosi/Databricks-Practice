```sql
%python

data = [
    ("2024-01-01", 120),
    ("2024-01-02", 150),
    ("2024-01-03", 130),
    ("2024-01-04", 180),
    ("2024-01-05", 200),
    ("2024-01-06", 170),
    ("2024-01-07", 220),
    ("2024-01-08", 210),
    ("2024-01-09", 190),
    ("2024-01-10", 230)
]

column = ["date", "sales"]

df = spark.createDataFrame(data, column)

display(df)
```

```sql
%python
df.createOrReplaceTempView("df_temp_view")
```

```sql
select * from df_temp_view
```

```sql
select
  date,
  sales,
  sum(sales) over(order by date) as cumulative_sales,
  lag(sales) over(order by date) as previous_day_sales,
  lead(sales) over(order by date) as next_day_sales
from
  df_temp_view
```

```sql
%python
from pyspark.sql import Window
from pyspark.sql.functions import col, sum, lag, lead

window_spec = Window.orderBy(col("date"))

df_final = (df
            .withColumn("cumulative_sales", 
                       sum(col("sales")).over(window_spec))
            .withColumn("previous_day_sales",
                       lag(col("sales")).over(window_spec))
            .withColumn("next_day_sales",
                       lead(col("sales")).over(window_spec))
)

display(df_final)
```
