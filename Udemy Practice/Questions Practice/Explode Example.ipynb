```sql
%python
from pyspark.sql.functions import explode, col

data = [
  (1, ["Pen", "Mobile" , "Laptop", "Book"]),
  (2, ["Bike", "Book"]),
  (3, ["Glass", "Watch", "Pen"]),
  (4, ["Bike", "Glass"]),
  (5, ["Laptop", "Bike", "Glass"]),
  (6, ["Pen", "Bike", "Glass"]),
  (7, ["Pen", "Mobile", "Glass", "Laptop", "Book"]),
  (8, ["Bike", "Pen", "Book"]),
  (9, ["Bike", "Glass", "Pen"]),
  (10, ["Laptop", "Mobile", "Book", "Glass", "Pen"])
]

column = ["id","product"]
df = spark.createDataFrame(data, column)

df_explode = (df
              .withColumn("product_purchase", explode(df.product))
              .select(col("id"), col("product_purchase"))
)
display(df_explode)
```

```sql
%python

from pyspark.sql.functions import coalesce, col, lit

data = (
    (1, "Alice Johnson", None, None),
    (2, None, "Bob Smith", None),
    (3, None, None, "Charlie Brown"),
    (4, "Diana Prince", "Diana P.", "D. Prince"),
    (5, None, "Evan Miller", "E. Miller"),
    (6, None, None, None),
    (7, "Grace Lee", None, "Grace L."),
    (8, None, None, "Henry Ford")
)

column = ["id", "name1", "name2", "name3"]
df = spark.createDataFrame(data, column)

display(df)

df_final = (df
            .withColumn("final_name", coalesce(col("name1"), col("name2"), col("name3"), lit("Unknown")))

)
display(df_final)
```
