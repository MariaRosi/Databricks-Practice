```sql
%python
customers_data = (
    (1, "Alice", "alice@gmail.com", "UK"),
    (2, "Bob", "bob@yahoo.com", "USA"),
    (3, "Charlie", "charlie@gmail.com", "UK"),
    (4, "Diana", "diana@outlook.com", "India"),
    (5, "Evan", "evan@gmail.com", "Canada"),
    (6, "Frank", "frank@yahoo.com", "USA"),
    (7, "Grace", "grace@gmail.com", "UK")
)

customers_columns = ["cust_id", "cust_name", "cust_email", "cust_country"]

orders_data = (
    (101, 1, "2024-01-10", 250.00),
    (102, 1, "2024-01-15", 120.00),
    (103, 2, "2024-02-01", 450.00),
    (104, 3, "2024-02-05", 300.00),
    (105, 3, "2024-02-10", 180.00),
    (106, 4, "2024-03-01", 600.00),
    # No matching customers
    (107, 8, "2024-03-04", 75.00),
    (108, 9, "2024-03-06", 220.00),
    (109, 10, "2024-03-08", 90.00),
    (110, 11, "2024-03-10", 150.00)
)

orders_columns = ["order_id", "cust_id", "order_date", "order_value"]

df_customers = spark.createDataFrame(customers_data, customers_columns)
df_orders = spark.createDataFrame(orders_data, orders_columns)

display(df_customers)
display(df_orders)
```

```sql
%python
display(df_customers
    .join(df_orders, df_customers.cust_id == df_orders.cust_id, "left")
    .filter(df_orders.order_id.isNull())
    .select(df_customers.cust_id, df_customers.cust_name, df_orders.order_id)
)
```

```sql
%python
display(df_customers
  .join(df_orders, df_customers.cust_id == df_orders.cust_id, "inner")
  .select(df_customers.cust_id, df_customers.cust_name, df_orders.order_id, df_orders.order_date)
)
```

```sql
%python
df_customers.createOrReplaceTempView("cust_temp_view")
df_orders.createOrReplaceTempView("orders_temp_view")

```

```sql
select * from cust_temp_view;

select * from orders_temp_view;
```

```sql
select
  c.cust_id,
  c.cust_name,
  o.order_id
from cust_temp_view c
left join orders_temp_view o
on c.cust_id = o.cust_id
where o.order_date is null
```

```sql
select
  c.cust_id,
  c.cust_name,
  o.order_id,
  o.order_date
from cust_temp_view c
inner join orders_temp_view o
on c.cust_id = o.cust_id
```
