```sql
%python

ata = [
    (1, "Alice", "Laptop"),
    (1, "Alice", "Mouse"),
    (1, "Alice", "Keyboard"),
    (1, "Alice", "Mouse"),
    (2, "Bob", "Phone"),
    (2, "Bob", "Headphones"),
    (2, "Bob", "Phone"),
    (3, "Charlie", "Tablet"),
    (3, "Charlie", "Tablet"),
    (3, "Charlie", "Stylus")
]

column = ["id", "name", "product"]

df = spark.createDataFrame(data=ata, schema=column)

display(df)
```

```sql
%python
df.createOrReplaceTempView("df_temp_view")
```

```sql
select
  id,
  name,
  collect_set(product) as products
from
  df_temp_view
group by id, name
```

```sql
%python
from pyspark.sql.functions import col, collect_set

df_final = (df
            .groupBy(col("id"), col("name"))
            .agg(collect_set(col("product")).alias("products"))
)

display(df_final)
```
