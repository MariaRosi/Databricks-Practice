```sql
%python
data = (
    (1, "john.doe@gmail.com"),
    (2, "jane.smith@yahoo.com"),
    (3, "mike.brown@outlook.com"),
    (1, "john.doe@gmail.com"),
    (5, "alice.wong@gmail.com"),
    (2, "jane.smith@yahoo.com"),
    (7, "robert.king@company.com"),
    (3, "mike.brown@outlook.com"),
    (9, "susan.lee@gmail.com"),
    (5, "ALICE.WONG@gmail.com"),
    (11, "emma.jones@yahoo.com"),
    (7, "robert.king@company.com"),
    (13, "paul.adams@gmail.com"),
    (11, "emma.jones@yahoo.com"),
    (11, "emma.jones@yahoo.com")
)

column = ["id","email"]

df = spark.createDataFrame(data, column)
display(df)
```

```sql
%python
from pyspark.sql.functions import count, collect_list

display(
    df.groupBy("email")
      .agg(
          count("email").alias("email_count"),
          collect_list("id").alias("ids")
      )
      .filter("email_count > 1")
)
```

```sql
%python
df.distinct().display()
```

```sql
%python
df.createOrReplaceTempView("df_temp_view")
```

```sql
select * from df_temp_view
```

```sql
select
  email,
  count(*) as email_count
from df_temp_view
group by email
having email_count > 1
```

```sql
with cte as
(
  select
    id,
    email,
    row_number() over(partition by email order by id) as rn
  from df_temp_view
)
select * from cte
where rn >  1
```
