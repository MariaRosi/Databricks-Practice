```sql
%python
data = [
  (1,"Alan",'{"city":"NY","state":"NY","zip":"10001"}'),
  (2,"Bob",'{"city":"LA","state":"CA","zip":"90001"}'),
  (3,"Charlie",'{"city":"SF","state":"CA","zip":"94101"}'),
  (4,"Dana",'{"city":"NY","state":"NY","zip":"10002"}'),
  (5,"Eve",'{"city":"LA","state":"CA","zip":"90002"}'),
  (6,"Frank",'{"city":"SF","state":"CA","zip":"94102"}')
]
column = ["id", "name", "address"]
df = spark.createDataFrame(data, column)

df.printSchema()
display(df)
```

```sql
%python
df.createOrReplaceTempView("df_temp_view")
```
