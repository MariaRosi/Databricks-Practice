```python
data = [
    ("Alice", "London", 12),
    ("Bob", "Manchester", 17),
    ("Charlie", "Birmingham", 23),
    ("Diana", "Leeds", 35),
    ("Edward", "Bristol", 45),
    ("Fiona", "Oxford", 62),
    ("George", "Reading", 28),
    ("Hannah", "Cambridge", 15),
    ("Ian", "Liverpool", 70),
    ("Julia", "Nottingham", 19)
]

columns = ["name", "city", "age"]

df = spark.createDataFrame(data, columns)
df.display()
```

```python
def age_grp(age):
    if age is None:
        return 'Unknown'
    elif age < 18:
        return 'MINOR'
    elif age > 18 and age < 50:
        return 'ADULT'
    else:
        return 'SENIOR'
    
age_grp(10)
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

age_grp_udf = udf(age_grp, StringType())
```

```python
df.withColumn("age_group", age_grp_udf("age")).show()
```
