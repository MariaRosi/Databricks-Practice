```sql
insert into table udemy_practice_catalog.sql_practice.employee select * from employee_temp_view;

select * from udemy_practice_catalog.sql_practice.employee
```

```sql
select
dept_id,
sum(salary) as total_salary
from udemy_practice_catalog.sql_practice.employee
group by dept_id
order by total_salary desc
```

```sql
select
dept_id,
sum(salary) as total_salary
from udemy_practice_catalog.sql_practice.employee
group by dept_id
having total_salary > 350000
order by total_salary desc
```

```sql

with cte as
(
  select
  *,
  row_number() over(partition by dept_id order by salary desc) as rn
  from udemy_practice_catalog.sql_practice.employee
)
select * from cte where rn = 2
order by salary desc
```

```sql
select dept_id, max(salary) as second_highest_salary
from udemy_practice_catalog.sql_practice.employee e1
where salary < (
  select max(salary)
  from udemy_practice_catalog.sql_practice.employee e2
  where e2.dept_id = e1.dept_id
)
group by dept_id
```

```sql
select
max(salary) as second_highest_salary
from udemy_practice_catalog.sql_practice.employee
where salary < (
  select max(salary) from udemy_practice_catalog.sql_practice.employee
)
```

```sql
select salary as second_highest_salary
from udemy_practice_catalog.sql_practice.employee
order by salary desc
limit 1 offset 1
```

```sql
select salary as second_highest_salary
from 
(
  select
  salary,
  row_number() over(order by salary desc) as rn
  from udemy_practice_catalog.sql_practice.employee
) where rn = 2
```
