```sql
-- Parameter in the where clause
select
  *
from samples.bakehouse.sales_customers
where gender = :param_gender and continent = :param_continent;

-- parameter for column select
select
  identifier(:param_column1),
  identifier(:param_column2)
from samples.bakehouse.sales_customers;

-- case statement
select
  *,
  case
    when gender = 'female' then 'F'
    when gender = 'male' then 'M'
    else 'Unknown'
  end as gender_short
from samples.bakehouse.sales_customers;

-- query snippet - case_gender_query_snippet
select
  *,
    case
      when gender = 'female' then 'F'
      when gender = 'male' then 'M'
      else 'Unknown'
    end as gender_short
from samples.bakehouse.sales_customers;

-- join
with cte as (
  select
    *
  from samples.bakehouse.sales_franchises as f
  join samples.bakehouse.sales_transactions as t
  on f.franchiseID = t.franchiseID
)
select
  country,
  product,
  sum(totalPrice) as totalPrice
from cte
group by country, product
order by country, totalPrice desc;

-- top 3 products with multiple cte
with cte_franchises_transactions_join as (
  select
    *
  from samples.bakehouse.sales_franchises as f
  join samples.bakehouse.sales_transactions as t
  on f.franchiseID = t.franchiseID
),
cte_group_by_product as (
  select
      product,
      sum(totalPrice) as TotalPrice
  from cte_franchises_transactions_join
  group by product
),
cte_tops_total_price as (
  select
    *,
    row_number() over(order by TotalPrice desc) as rn
  from cte_group_by_product
  order by TotalPrice desc
)
select
   *
from cte_tops_total_price
where rn <= 3;

-- with limit
with cte as (
  select
    *
  from samples.bakehouse.sales_franchises as f
  join samples.bakehouse.sales_transactions as t
  on f.franchiseID = t.franchiseID
)
select
   product,
   sum(totalPrice) as TotalPrice
from cte
group by product
order by TotalPrice desc
limit 3;










```
