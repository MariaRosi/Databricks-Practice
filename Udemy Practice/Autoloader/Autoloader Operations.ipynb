```python
df = spark.readStream.format("cloudFiles") \
    .option("cloudFiles.format", "json") \
    .option("cloudFiles.schemaLocation", "/Volumes/udemy_practice_catalog/autoloader/bronze/destination/checkpoint/") \
    .option("cloudFiles.schemaEvolutionMode", "rescue") \
    .option("multiLine", "true") \
    .load("/Volumes/udemy_practice_catalog/autoloader/bronze/raw/")
```

```python
df.writeStream.format("delta") \
     .outputMode("append") \
     .option("checkpointLocation", "/Volumes/udemy_practice_catalog/autoloader/bronze/destination/checkpoint/") \
     .trigger(once = True) \
     .start("/Volumes/udemy_practice_catalog/autoloader/bronze/destination/data/")
```
