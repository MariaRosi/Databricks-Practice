```python
data = [
    (1, "Alice",   "IT",      90000),
    (2, "Bob",     "IT",      85000),
    (3, "Charlie", "IT",      90000),
    (4, "Diana",   "HR",      70000),
    (5, "Edward",  "HR",      65000),
    (6, "Fiona",   "HR",      70000),
    (7, "George",  "Finance", 95000),
    (8, "Hannah",  "Finance", 88000),
    (9, "Ian",     "Finance", 88000),
    (10, "Julia",  "Sales",   60000),
    (11, "Kevin",  "Sales",   62000),
    (12, "Laura",  "Sales",   60000)
]

columns = ["id", "name", "dept", "salary"]

df = spark.createDataFrame(data, columns)
df.display()
```

```python
df.createOrReplaceTempView("emp_temp_view")
```

```python
%sql
SELECT * FROM emp_temp_view
```

```python
%sql
SELECT
  *,
  ROW_NUMBER() OVER(PARTITION BY dept ORDER BY salary DESC) as rn
FROM
  emp_temp_view
```

```python
%sql
WITH emp_cte AS (
    SELECT
    *,
    ROW_NUMBER() OVER(PARTITION BY dept ORDER BY salary DESC) as rn
  FROM
    emp_temp_view
)
select * from emp_cte
where rn = 1
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, col

w = Window.partitionBy("dept").orderBy(col("salary").desc())

df.withColumn("rn", row_number().over(w)).display()
```

```python
from pyspark.sql.functions import rank

w = Window.partitionBy("dept").orderBy(col("salary").desc())

df.withColumn("rn", rank().over(w)).display()
```

```python
from pyspark.sql.functions import dense_rank

w = Window.partitionBy("dept").orderBy(col("salary").desc())

df.withColumn("rn", dense_rank().over(w)).display()
```

```python
from pyspark.sql.functions import lag

w = Window.partitionBy("dept").orderBy(col("salary").desc())

df.withColumn("rn", lag("salary", 2).over(w)).display()
```

```python
from pyspark.sql.functions import lead

w = Window.partitionBy("dept").orderBy(col("salary").desc())

df.withColumn("rn", lead("salary").over(w)).display()
```
