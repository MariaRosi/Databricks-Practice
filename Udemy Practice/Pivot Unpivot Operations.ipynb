### Pivot - Row To Colum
### UnPivot - Column To Row

```python
data = [
    ["East",  "Laptop", 10],
    ["East",  "Tablet", 15],
    ["East",  "Mobile", 20],
    ["West",  "Laptop", 7],
    ["West",  "Tablet", 12],
    ["West",  "Mobile", 18],
    ["North", "Laptop", 5],
    ["North", "Tablet", 9],
    ["North", "Mobile", 14],
    ["South", "Laptop", 8],
    ["South", "Tablet", 11],
    ["South", "Mobile", 16],
    ["East",  "Laptop", 6],
    ["West",  "Mobile", 4],
    ["South", "Tablet", 5]
]

columns = ["region", "product", "quantity"]
df = spark.createDataFrame(data, columns)

display(df)
```

```python
from pyspark.sql.functions import sum

display(df.groupBy("region").pivot("product").agg(sum("quantity")))
```

```python
display(df.groupBy("product").pivot("region").agg(sum("quantity")))
```

```python
from pyspark.sql.functions import expr

df1 = df.groupBy("region").pivot("product").agg(sum("quantity"))
unpivot_df = df1.select(
            "region",
            expr("""
                stack(
                    3,
                    'Laptop', Laptop,
                    'Tablet', Tablet,
                    'Mobile', Mobile
                ) as (product, quantity)    
            """
            )
        )

display(unpivot_df)
```
