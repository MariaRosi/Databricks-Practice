### **distinct And dropDuplicates**

```python

data = [
    (1, "Alice", 25),
    (2, "Bob", 30),
    (3, "Charlie", 35),
    (4, " David", 40),
    (5, "Eve", 28),
    (2, "Bob", 30),           # duplicate row
    (3, "Charlie", 35),       # duplicate row
    (6, "Alice", 25),         # duplicate values in 'name' and 'age'
    (7, "Eve", 28)            # duplicate values in 'name' and 'age'
]

header = ["id", "name", "age"]

df = spark.createDataFrame(data, header)

display(df)
```

```python
display(df.distinct())
```

```python
display(df.dropDuplicates())
```

```python
display(df.dropDuplicates(['name', 'age']))
```

```python
display(df.dropDuplicates(['name']))
```

### **sort and orderby**

```python
df = spark.read.parquet("/Volumes/udemy_practice_catalog/udemy_practice_schema/udemy_practice_volume/orders/orders.parquet")

display(df)
```

```python
from pyspark.sql.functions import col
display(df.sort("customer_name").filter(col("customer_name").isNotNull()))
```

```python
display(df.sort("customer_name").filter(df.customer_name.isNotNull()))
```

```python
display(df.orderBy("customer_name").filter("customer_name IS NOT NULL"))
```

```python
display(df.orderBy("order_id", ascending=False).filter("customer_name IS NOT NULL AND order_id IS NOT NULL"))
```

```python
display(df.orderBy(df.order_id.desc()).filter(df.order_id.isNotNull()).filter(df.customer_name.isNotNull()))
```
